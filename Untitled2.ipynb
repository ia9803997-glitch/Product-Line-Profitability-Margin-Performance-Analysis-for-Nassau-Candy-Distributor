```python

import pandas as pd

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)

df.head()
```

```python

print("Column Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python

# 1. Convert date columns into datetime format
df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

# 2. Check date conversion errors
print("Invalid Order Dates:", df["Order Date"].isna().sum())
print("Invalid Ship Dates:", df["Ship Date"].isna().sum())

# 3. Check financial values
print("\nZero Sales:", (df["Sales"] == 0).sum())
print("Negative Sales:", (df["Sales"] < 0).sum())
print("Negative Cost:", (df["Cost"] < 0).sum())
print("Negative Gross Profit:", (df["Gross Profit"] < 0).sum())
print("Zero or Negative Units:", (df["Units"] <= 0).sum())

# 4. Check whether gross profit matches Sales - Cost
df["Profit Difference"] = (
    df["Gross Profit"] - (df["Sales"] - df["Cost"])
).round(2)

print(
    "\nRows where Gross Profit differs from Sales - Cost:",
    (df["Profit Difference"].abs() > 0.01).sum()
)

# 5. Display the first five records
df.head()
```

```python

# Reload the original CSV to recover date values
df = pd.read_csv(file_name)

# Convert dates using day-first format
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True,
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True,
    errors="coerce"
)

# Check the results
print("Invalid Order Dates:", df["Order Date"].isna().sum())
print("Invalid Ship Dates:", df["Ship Date"].isna().sum())

print("\nOrder Date Range:")
print(df["Order Date"].min(), "to", df["Order Date"].max())

print("\nShip Date Range:")
print(df["Ship Date"].min(), "to", df["Ship Date"].max())

df[["Order Date", "Ship Date"]].head()
```

```python

# Reload the original dates without converting them
raw_df = pd.read_csv(file_name, dtype={"Order Date": str, "Ship Date": str})

print("Original date values:")
print(raw_df[["Order ID", "Order Date", "Ship Date"]].head(10).to_string(index=False))

# Convert dates using day-first format for comparison
orders = pd.to_datetime(raw_df["Order Date"], dayfirst=True, errors="coerce")
ships = pd.to_datetime(raw_df["Ship Date"], dayfirst=True, errors="coerce")

print("\nShipping delay in days:")
print((ships - orders).head(10))
```

```python

# Convert original date columns consistently
order_dates = pd.to_datetime(
    raw_df["Order Date"], dayfirst=True, errors="coerce"
)

ship_dates = pd.to_datetime(
    raw_df["Ship Date"], dayfirst=True, errors="coerce"
)

# Calculate shipping delay
delay_days = (ship_dates - order_dates).dt.days

print("Total records:", len(raw_df))
print("Shipping before order:", (delay_days < 0).sum())
print("Same-day shipping:", (delay_days == 0).sum())
print("Shipping within 7 days:", ((delay_days > 0) & (delay_days <= 7)).sum())
print("Shipping after 30 days:", (delay_days > 30).sum())

print("\nShipping delay summary:")
print(delay_days.describe())

print("\nSample records:")
print(raw_df[["Order ID", "Order Date", "Ship Date"]].head(10))
```

```python

# Create a data-quality check column
df["Shipping Delay Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

df["Date Quality Flag"] = df["Shipping Delay Days"].apply(
    lambda days: "Review Required" if days > 30 or days < 0
    else "Check Passed"
)

print(df["Date Quality Flag"].value_counts())

print("\nDate quality examples:")
print(
    df[["Order ID", "Order Date", "Ship Date",
        "Shipping Delay Days", "Date Quality Flag"]].head()
)
```

```python

# Calculate overall business performance

total_sales = df["Sales"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Gross Profit"].sum()
total_units = df["Units"].sum()

gross_margin = (total_profit / total_sales) * 100

profit_per_unit = total_profit / total_units

print("===== BUSINESS PERFORMANCE =====")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Cost: ${total_cost:,.2f}")
print(f"Total Gross Profit: ${total_profit:,.2f}")
print(f"Total Units Sold: {total_units:,}")
print(f"Gross Margin: {gross_margin:.2f}%")
print(f"Average Profit per Unit: ${profit_per_unit:.2f}")
```

```python

# Group data product-wise
product_analysis = (
    df.groupby(["Product ID", "Product Name"], as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Cost=("Cost", "sum"),
        Total_Profit=("Gross Profit", "sum"),
        Total_Units=("Units", "sum")
    )
)

# Calculate product profitability metrics
product_analysis["Gross_Margin_%"] = (
    product_analysis["Total_Profit"]
    / product_analysis["Total_Sales"] * 100
).round(2)

product_analysis["Profit_Per_Unit"] = (
    product_analysis["Total_Profit"]
    / product_analysis["Total_Units"]
).round(2)

# Rank products by total gross profit
top_10_products = product_analysis.sort_values(
    by="Total_Profit", ascending=False
).head(10)

print("TOP 10 PROFITABLE PRODUCTS")

display(
    top_10_products[
        [
            "Product Name",
            "Total_Sales",
            "Total_Profit",
            "Gross_Margin_%",
            "Total_Units",
            "Profit_Per_Unit"
        ]
    ].reset_index(drop=True)
)
```

```python

# Find the 10 products with the lowest gross margins
lowest_margin_products = product_analysis.sort_values(
    by="Gross_Margin_%",
    ascending=True
).head(10)

print("10 LOWEST MARGIN PRODUCTS")

display(
    lowest_margin_products[
        [
            "Product Name",
            "Total_Sales",
            "Total_Cost",
            "Total_Profit",
            "Gross_Margin_%",
            "Total_Units"
        ]
    ].reset_index(drop=True)
)
```

```python

division_analysis = (
    df.groupby("Division", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Cost=("Cost", "sum"),
        Total_Profit=("Gross Profit", "sum"),
        Total_Units=("Units", "sum")
    )
)

division_analysis["Gross_Margin_%"] = (
    division_analysis["Total_Profit"]
    / division_analysis["Total_Sales"] * 100
).round(2)

division_analysis = division_analysis.sort_values(
    "Total_Profit", ascending=False
)

print("DIVISION-WISE PROFITABILITY")
display(division_analysis)
```

```python

region_analysis = (
    df.groupby("Region", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Cost=("Cost", "sum"),
        Total_Profit=("Gross Profit", "sum"),
        Total_Units=("Units", "sum")
    )
)

region_analysis["Gross_Margin_%"] = (
    region_analysis["Total_Profit"]
    / region_analysis["Total_Sales"] * 100
).round(2)

region_analysis = region_analysis.sort_values(
    "Total_Profit", ascending=False
)

print("REGION-WISE PROFITABILITY")
display(region_analysis)
```

```python

import matplotlib.pyplot as plt

# Chart 1: Profit by Division
plt.figure(figsize=(8, 5))
plt.bar(
    division_analysis["Division"],
    division_analysis["Total_Profit"]
)
plt.title("Gross Profit by Division")
plt.xlabel("Division")
plt.ylabel("Gross Profit ($)")
plt.tight_layout()
plt.show()

# Chart 2: Profit by Region
plt.figure(figsize=(8, 5))
plt.bar(
    region_analysis["Region"],
    region_analysis["Total_Profit"]
)
plt.title("Gross Profit by Region")
plt.xlabel("Region")
plt.ylabel("Gross Profit ($)")
plt.tight_layout()
plt.show()

# Chart 3: Top 10 Products by Gross Profit
plt.figure(figsize=(10, 6))
plt.barh(
    top_10_products["Product Name"].iloc[::-1],
    top_10_products["Total_Profit"].iloc[::-1]
)
plt.title("Top 10 Products by Gross Profit")
plt.xlabel("Gross Profit ($)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python

# Monthly Sales Trend Analysis

df["Order Month"] = df["Order Date"].dt.to_period("M")

monthly_analysis = (
    df.groupby("Order Month", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Gross Profit", "sum")
    )
)

monthly_analysis["Order Month"] = (
    monthly_analysis["Order Month"].astype(str)
)

print("Monthly Sales and Profit Analysis:")
display(monthly_analysis.head(12))

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(
    monthly_analysis["Order Month"],
    monthly_analysis["Total_Sales"],
    marker="o"
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# STEP 11: Region-wise Sales vs Gross Profit Analysis

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Region-wise summary
region_summary = (
    df.groupby("Region", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Gross Profit", "sum"),
        Total_Units=("Units", "sum")
    )
)

# 2. Calculate gross profit margin
region_summary["Profit_Margin_%"] = (
    region_summary["Total_Profit"]
    / region_summary["Total_Sales"] * 100
).round(2)

# 3. Sort by highest profit
region_summary = region_summary.sort_values(
    "Total_Profit", ascending=False
)

# 4. Display results
print("Region-wise Sales and Profit Summary")
display(region_summary.round(2))

# 5. Compare sales and profit visually
x = np.arange(len(region_summary))
width = 0.35

plt.figure(figsize=(10, 5))

plt.bar(
    x - width / 2,
    region_summary["Total_Sales"],
    width,
    label="Total Sales"
)

plt.bar(
    x + width / 2,
    region_summary["Total_Profit"],
    width,
    label="Gross Profit"
)

plt.xticks(x, region_summary["Region"])
plt.title("Region-wise Sales vs Gross Profit")
plt.xlabel("Region")
plt.ylabel("Amount ($)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Step 12: Profit Concentration (Pareto) Analysis

# Product-wise sales and profit
pareto = product_analysis.copy()

# Sales concentration
pareto = pareto.sort_values("Total_Sales", ascending=False)
pareto["Cumulative_Sales_%"] = (
    pareto["Total_Sales"].cumsum() /
    pareto["Total_Sales"].sum() * 100
)

# Profit concentration
pareto = pareto.sort_values("Total_Profit", ascending=False)
pareto["Cumulative_Profit_%"] = (
    pareto["Total_Profit"].cumsum() /
    pareto["Total_Profit"].sum() * 100
)

# Number of products contributing to 80% of sales
sales_sorted = pareto.sort_values("Total_Sales", ascending=False)
sales_count = (
    sales_sorted["Cumulative_Sales_%"] < 80
).sum() + 1

# Number of products contributing to 80% of profit
profit_sorted = pareto.sort_values("Total_Profit", ascending=False)
profit_count = (
    profit_sorted["Cumulative_Profit_%"] < 80
).sum() + 1

print("Products contributing to 80% of total sales:", sales_count)
print("Products contributing to 80% of total profit:", profit_count)

print("\nTop products by profit concentration:")
display(profit_sorted[
    ["Product Name", "Total_Sales", "Total_Profit",
     "Cumulative_Profit_%"]
].head(10).round(2))
```

```python
# Step 15: Margin Risk Analysis

margin_risk = product_analysis.copy()

margin_risk["Margin_Risk"] = margin_risk["Gross_Margin_%"].apply(
    lambda margin: "High Risk" if margin < 20
    else "Medium Risk" if margin < 50
    else "Low Risk"
)

print("Product Margin Risk Analysis")

display(
    margin_risk[
        [
            "Product Name",
            "Total_Sales",
            "Total_Profit",
            "Gross_Margin_%",
            "Margin_Risk"
        ]
    ].sort_values("Gross_Margin_%").round(2)
)

print("\nProducts requiring pricing/cost review:")

display(
    margin_risk[
        margin_risk["Margin_Risk"] == "High Risk"
    ][
        ["Product Name", "Total_Sales", "Total_Profit", "Gross_Margin_%"]
    ].round(2)
)
```
